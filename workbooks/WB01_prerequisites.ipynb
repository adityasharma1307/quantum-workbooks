# Part A: Theory notes

## 1.1 Why quantum computing needs new maths

Classical computers store bits. A bit is 0 or 1, and you can read it as often as you like without changing it. Quantum systems behave differently, and two classic experiments show how.

### Polarised light

Light has a polarisation direction. A polarising filter (like the film in some sunglasses) lets through light polarised along its axis and blocks light polarised at right angles to it. For light polarised at angle $\theta$ to the filter axis, the transmitted intensity follows **Malus's law**:

$$
I = I_0 \cos^2\theta .
$$

Now try three filters. Put a vertical filter and a horizontal filter in a row. No light gets through, since $\cos^2 90^\circ = 0$. Now slide a third filter at $45^\circ$ *between* them. Light comes out again. If unpolarised light of intensity $I_0$ enters, the first filter passes $I_0/2$, the middle one passes half of that, and the last one passes half again:

$$
I_{\text{out}} = \frac{I_0}{2}\cdot\cos^2 45^\circ\cdot\cos^2 45^\circ = \frac{I_0}{8}.
$$

Adding a filter let *more* light through. Now turn the light down until single photons arrive one at a time. Each photon either passes a filter or it does not. The detector clicks or stays silent. What Malus's law gives you is a **probability**: a photon polarised at angle $\theta$ to the filter passes with probability $\cos^2\theta$. After it passes, it is polarised along the filter axis, whatever it was before.

### The Stern-Gerlach experiment

In 1922 Otto Stern and Walther Gerlach sent a beam of silver atoms through a non-uniform magnetic field. A tiny classical magnet would be deflected by a continuous range of amounts, giving a smear on the screen. Instead the atoms landed in **two spots**: "up" and "down". Now chain the devices. Select the "up" atoms from a vertical device, send them through a horizontal device, and then through a vertical device again. The final vertical device gives "up" and "down" in roughly equal numbers. The horizontal measurement wiped out the earlier vertical result.

### The ideas in plain words

These experiments suggest four ideas that the rest of the course makes precise.

- **States are vectors.** The state of a photon's polarisation, or of an atom's spin, is described by a vector. The vector holds *amplitudes*, which can be complex numbers.
- **Superposition** means the state vector is a combination of two basis states, such as "vertical" and "horizontal". It is a definite state of its own. It is *not* a hidden list of answers waiting to be read off.
- **Measurement** gives one outcome from a fixed set (pass or block, up or down). Which outcome you get is random. The probabilities come from the amplitudes.
- **Measurement changes the state.** After a measurement the system is left in the state that matches the outcome.

To work with these ideas we need complex numbers, vectors, matrices and probability. That is the rest of this workbook.

> **Note.** A qubit is any two-level quantum system: photon polarisation, the spin of an electron or atom, or two energy levels of a superconducting circuit. The maths in this course is the same for all of them.

## 1.2 Complex numbers

A **complex number** is $z = a + ib$, where $a$ and $b$ are real numbers and $i^2 = -1$. We call $a = \mathrm{Re}(z)$ the real part and $b = \mathrm{Im}(z)$ the imaginary part.

**Arithmetic.** Add parts separately, and multiply out brackets using $i^2=-1$:

$$
(a+ib)+(c+id) = (a+c) + i(b+d), \qquad (a+ib)(c+id) = (ac-bd) + i(ad+bc).
$$

**Complex conjugate.** Flip the sign of the imaginary part: $z^\ast = a - ib$.

**Modulus.** The length of $z$ in the complex plane:

$$
\lvert z\rvert = \sqrt{a^2+b^2}, \qquad \lvert z\rvert^2 = z^\ast z .
$$

The modulus squared $\lvert z\rvert^2$ will become a probability in Workbook 2, so get comfortable with it now.

**Division.** Multiply top and bottom by the conjugate of the denominator: $\dfrac{z}{w} = \dfrac{z\,w^\ast}{\lvert w\rvert^2}$.

### Worked example 1.1

Let $z_1 = 3+4i$ and $z_2 = 1-2i$.

- Sum: $z_1 + z_2 = 4 + 2i$.
- Product: $z_1 z_2 = 3 - 6i + 4i - 8i^2 = 11 - 2i$.
- Conjugate and modulus: $z_1^\ast = 3-4i$ and $\lvert z_1\rvert = \sqrt{9+16} = 5$.
- Quotient: $\dfrac{z_1}{z_2} = \dfrac{(3+4i)(1+2i)}{1^2+2^2} = \dfrac{-5+10i}{5} = -1+2i$.

### Polar form and Euler's formula

Any complex number can be written using its length $r = \lvert z\rvert$ and its angle $\theta$ from the positive real axis. **Euler's formula** connects the two forms:

$$
e^{i\theta} = \cos\theta + i\sin\theta, \qquad z = re^{i\theta}.
$$

Some values to remember: $e^{i0}=1$, $e^{i\pi/2}=i$, $e^{i\pi}=-1$ and $e^{i3\pi/2}=-i$. Also $\lvert e^{i\theta}\rvert = 1$ for every real $\theta$. A number of the form $e^{i\theta}$ is called a **phase**.

Polar form makes multiplication easy: lengths multiply and angles add.

$$
r_1e^{i\theta_1}\cdot r_2e^{i\theta_2} = r_1r_2\,e^{i(\theta_1+\theta_2)} .
$$

So multiplying by $i = e^{i\pi/2}$ rotates a number by $90^\circ$, and multiplying by a phase $e^{i\theta}$ rotates without changing the length.

### Worked example 1.2

Write $1+i$ in polar form. The length is $r = \sqrt{1^2+1^2} = \sqrt{2}$ and the angle is $\theta = \pi/4$, so $1+i = \sqrt{2}\,e^{i\pi/4}$. Check: $\sqrt2(\cos\frac{\pi}{4} + i\sin\frac{\pi}{4}) = \sqrt2\left(\frac{1}{\sqrt2}+\frac{i}{\sqrt2}\right) = 1+i$.

> **Common mistake.** Writing $\lvert z\rvert^2 = z^2$. For $z = 1+i$, $z^2 = 2i$, which is not even real. The correct rule is $\lvert z\rvert^2 = z^\ast z = (1-i)(1+i) = 2$.

> **Note.** Python writes the imaginary unit as `1j`, so $3+4i$ is `3+4j`. The letter `i` alone is just a variable name.

```python
import numpy as np

z1 = 3 + 4j
z2 = 1 - 2j
print("sum      :", z1 + z2)
print("product  :", z1 * z2)
print("conj(z1) :", np.conj(z1))
print("|z1|     :", abs(z1))
print("z1 / z2  :", z1 / z2)

# Polar form of 1 + i: length and angle
r, theta = abs(1 + 1j), np.angle(1 + 1j)
print("r =", round(r, 4), " theta/pi =", round(theta / np.pi, 4))
print("e^{i pi} =", np.round(np.exp(1j * np.pi), 10))
```

## 1.3 Vectors in Dirac notation

In quantum computing we write column vectors as **kets**, using a notation invented by Paul Dirac. A ket looks like $|\psi\rangle$ (read "ket psi"). For one qubit we use vectors with two complex entries, so they live in $\mathbb{C}^2$.

The two most important kets are the **computational basis** vectors:

$$
|0\rangle = \begin{pmatrix}1\\0\end{pmatrix}, \qquad |1\rangle = \begin{pmatrix}0\\1\end{pmatrix}.
$$

Every vector in $\mathbb{C}^2$ is a combination of these two:

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix}\alpha\\ \beta\end{pmatrix}, \qquad \alpha,\beta\in\mathbb{C}.
$$

Two more kets appear all the time. They are called "plus" and "minus":

$$
|+\rangle = \frac{1}{\sqrt2}\begin{pmatrix}1\\1\end{pmatrix}, \qquad |-\rangle = \frac{1}{\sqrt2}\begin{pmatrix}1\\-1\end{pmatrix}.
$$

**Bras.** The **bra** $\langle\psi|$ is the conjugate transpose of the ket: turn the column into a row and conjugate each entry.

$$
|\psi\rangle = \begin{pmatrix}\alpha\\ \beta\end{pmatrix} \quad\Longrightarrow\quad \langle\psi| = \begin{pmatrix}\alpha^\ast & \beta^\ast\end{pmatrix}.
$$

> **Note.** The label inside the ket is just a name. $|0\rangle$ is *not* the zero vector. The zero vector is written $0$ or $\mathbf{0}$, with no ket around it.

### Worked example 1.3

Let $|\psi\rangle = \frac{1}{\sqrt2}\left(|0\rangle + i|1\rangle\right)$. As a column vector it is $\frac{1}{\sqrt2}\begin{pmatrix}1\\ i\end{pmatrix}$, and its bra is $\langle\psi| = \frac{1}{\sqrt2}\begin{pmatrix}1 & -i\end{pmatrix}$. Notice the sign change on the $i$.

## 1.4 Inner products and norms

The **inner product** of two kets $|\phi\rangle$ and $|\psi\rangle$ is the bra of the first times the ket of the second. We write it $\langle\phi|\psi\rangle$ (a "bra-ket"):

$$
\langle\phi|\psi\rangle = \begin{pmatrix}\phi_0^\ast & \phi_1^\ast\end{pmatrix}\begin{pmatrix}\psi_0\\ \psi_1\end{pmatrix} = \phi_0^\ast\psi_0 + \phi_1^\ast\psi_1 .
$$

The result is a single complex number. Three properties matter:

1. **Swapping conjugates:** $\langle\psi|\phi\rangle = \langle\phi|\psi\rangle^\ast$.
2. **Positivity:** $\langle\psi|\psi\rangle = \lvert\psi_0\rvert^2 + \lvert\psi_1\rvert^2 \geq 0$, and it is $0$ only for the zero vector.
3. **Linearity in the ket:** $\langle\phi|(a|\psi\rangle + b|\chi\rangle) = a\langle\phi|\psi\rangle + b\langle\phi|\chi\rangle$.

The **norm** (length) of a ket is $\lVert\psi\rVert = \sqrt{\langle\psi|\psi\rangle}$. A ket with norm 1 is **normalised**. Two kets are **orthogonal** when $\langle\phi|\psi\rangle = 0$. A pair of normalised, orthogonal kets in $\mathbb{C}^2$ is an **orthonormal basis**. Both $\{|0\rangle,|1\rangle\}$ and $\{|+\rangle,|-\rangle\}$ are orthonormal bases.

### Worked example 1.4

Let $|a\rangle = \frac{1}{\sqrt2}\begin{pmatrix}1\\ i\end{pmatrix}$ and $|b\rangle = \frac{1}{\sqrt2}\begin{pmatrix}1\\ 1\end{pmatrix} = |+\rangle$.

$$
\langle b|a\rangle = \tfrac{1}{2}\left(1\cdot 1 + 1\cdot i\right) = \tfrac{1+i}{2}, \qquad \langle a|b\rangle = \tfrac{1}{2}\left(1\cdot1 + (-i)\cdot 1\right) = \tfrac{1-i}{2}.
$$

The two answers are conjugates, as property 1 says. Their modulus squared is $\lvert\tfrac{1+i}{2}\rvert^2 = \tfrac{2}{4} = \tfrac12$. In Workbook 2 you will see that this number is a measurement probability.

Also $\langle +|-\rangle = \frac12(1\cdot 1 + 1\cdot(-1)) = 0$, so $|+\rangle$ and $|-\rangle$ are orthogonal.

> **Common mistake.** Forgetting to conjugate the bra. In NumPy, `np.dot(a, b)` does *not* conjugate. Use `np.vdot(a, b)`, which conjugates its first argument and computes $\langle a|b\rangle$.

```python
a = np.array([1, 1j]) / np.sqrt(2)
b = np.array([1, 1]) / np.sqrt(2)

print("<b|a> =", np.round(np.vdot(b, a), 4))   # conjugates b
print("<a|b> =", np.round(np.vdot(a, b), 4))   # conjugates a
print("|<b|a>|^2 =", round(abs(np.vdot(b, a))**2, 4))
print("norm of a =", round(np.linalg.norm(a), 4))
print("np.dot(a, a) =", np.round(np.dot(a, a), 4), " <- wrong: no conjugate")
```

## 1.5 Matrices as operators

A $2\times2$ matrix acts on a ket by matrix multiplication and gives another ket. In quantum computing, matrices describe what we *do* to a state.

$$
\begin{pmatrix}a & b\\ c & d\end{pmatrix}\begin{pmatrix}\alpha\\ \beta\end{pmatrix} = \begin{pmatrix}a\alpha + b\beta\\ c\alpha + d\beta\end{pmatrix}.
$$

Here are five matrices you will meet again and again. In Workbook 3 they become quantum gates. For now they are just matrices.

$$
I = \begin{pmatrix}1&0\\0&1\end{pmatrix},\quad
X = \begin{pmatrix}0&1\\1&0\end{pmatrix},\quad
Y = \begin{pmatrix}0&-i\\ i&0\end{pmatrix},\quad
Z = \begin{pmatrix}1&0\\0&-1\end{pmatrix},\quad
H = \frac{1}{\sqrt2}\begin{pmatrix}1&1\\1&-1\end{pmatrix}.
$$

$X$, $Y$ and $Z$ are the **Pauli matrices**, and $H$ is the **Hadamard matrix**. Check by multiplying out: $X|0\rangle = |1\rangle$, $Z|1\rangle = -|1\rangle$, $H|0\rangle = |+\rangle$ and $H|1\rangle = |-\rangle$.

**Order matters.** Matrix products usually do not commute. For example

$$
XZ = \begin{pmatrix}0&-1\\1&0\end{pmatrix}, \qquad ZX = \begin{pmatrix}0&1\\-1&0\end{pmatrix} = -XZ .
$$

In $XZ|\psi\rangle$, the matrix nearest the ket acts first. So $Z$ acts first, then $X$.

**Outer products.** A ket times a bra, $|\psi\rangle\langle\phi|$, is a $2\times2$ matrix. For example $|0\rangle\langle 0| = \begin{pmatrix}1&0\\0&0\end{pmatrix}$, and $|0\rangle\langle 0| + |1\rangle\langle 1| = I$.

```python
I2 = np.eye(2)
X = np.array([[0, 1], [1, 0]])
Y = np.array([[0, -1j], [1j, 0]])
Z = np.array([[1, 0], [0, -1]])
H = np.array([[1, 1], [1, -1]]) / np.sqrt(2)
ket0, ket1 = np.array([1, 0]), np.array([0, 1])

print("H|0> =", np.round(H @ ket0, 4))
print("XZ =\n", X @ Z)
print("ZX =\n", Z @ X)
print("|0><0| + |1><1| =\n", np.outer(ket0, ket0) + np.outer(ket1, ket1))
```

Qiskit has these matrices built in. The `Operator` class turns a gate into its matrix, so you can check that Qiskit agrees with the definitions above.

```python
from qiskit.quantum_info import Operator
from qiskit.circuit.library import XGate, YGate, ZGate, HGate

for name, gate, M in [("X", XGate(), X), ("Y", YGate(), Y), ("Z", ZGate(), Z), ("H", HGate(), H)]:
    print(name, "matches:", np.allclose(Operator(gate).data, M))
```

## 1.6 The dagger, unitary and Hermitian matrices

The **conjugate transpose** of a matrix $A$, written $A^\dagger$ ("A dagger"), is found by transposing and then conjugating every entry:

$$
A = \begin{pmatrix}a&b\\c&d\end{pmatrix} \quad\Longrightarrow\quad A^\dagger = \begin{pmatrix}a^\ast&c^\ast\\b^\ast&d^\ast\end{pmatrix}.
$$

The bra is just the dagger of the ket: $\langle\psi| = (|\psi\rangle)^\dagger$. A useful rule is that the dagger reverses products: $(AB)^\dagger = B^\dagger A^\dagger$.

### Worked example 1.5

For $A = \begin{pmatrix}1 & 2i\\ 3 & 4+i\end{pmatrix}$, transposing gives $\begin{pmatrix}1&3\\2i&4+i\end{pmatrix}$, and conjugating gives $A^\dagger = \begin{pmatrix}1 & 3\\ -2i & 4-i\end{pmatrix}$.

### Two special families

A matrix $U$ is **unitary** if

$$
U^\dagger U = UU^\dagger = I .
$$

Unitary matrices keep lengths and inner products the same: if $|\psi'\rangle = U|\psi\rangle$ and $|\phi'\rangle = U|\phi\rangle$, then $\langle\phi'|\psi'\rangle = \langle\phi|U^\dagger U|\psi\rangle = \langle\phi|\psi\rangle$. They are also reversible, since $U^{-1} = U^\dagger$. In this course, **every quantum gate is a unitary matrix.**

A matrix $A$ is **Hermitian** if

$$
A^\dagger = A .
$$

The diagonal entries of a Hermitian matrix are real, and (as you will show in the exercises) its eigenvalues are real. Hermitian matrices describe **measurable quantities** (observables).

Some matrices are both. $X$, $Y$, $Z$ and $H$ are all unitary *and* Hermitian. Others are only one of the two. The phase matrix $S = \begin{pmatrix}1&0\\0&i\end{pmatrix}$ is unitary but not Hermitian, because $S^\dagger = \begin{pmatrix}1&0\\0&-i\end{pmatrix} \neq S$. The matrix $\begin{pmatrix}1&1\\1&1\end{pmatrix}$ is Hermitian but not unitary.

> **Common mistake.** Thinking "unitary" means "has determinant 1" or "entries have modulus at most 1". The test is $U^\dagger U = I$. Check it by multiplying, or by checking that the columns form an orthonormal basis.

```python
def dagger(M):
    return M.conj().T

def is_unitary(M):
    return np.allclose(dagger(M) @ M, np.eye(len(M)))

def is_hermitian(M):
    return np.allclose(dagger(M), M)

S = np.array([[1, 0], [0, 1j]])
J = np.array([[1, 1], [1, 1]])
for name, M in [("H", H), ("Y", Y), ("S", S), ("[[1,1],[1,1]]", J)]:
    print(f"{name:14s} unitary={is_unitary(M)!s:5s} hermitian={is_hermitian(M)}")
```

## 1.7 Eigenvalues and eigenvectors of 2x2 matrices

A nonzero ket $|v\rangle$ is an **eigenvector** of $A$ with **eigenvalue** $\lambda$ if

$$
A|v\rangle = \lambda|v\rangle .
$$

The matrix only stretches the eigenvector (by a factor $\lambda$, which may be complex). It does not change its direction.

**Recipe for a 2x2 matrix.** The eigenvalues solve $\det(A - \lambda I) = 0$. For a $2\times2$ matrix this is a quadratic:

$$
\lambda^2 - (\operatorname{tr}A)\,\lambda + \det A = 0,
$$

where $\operatorname{tr}A = a + d$ is the trace (sum of the diagonal) and $\det A = ad - bc$. For each $\lambda$, solve $(A-\lambda I)|v\rangle = 0$ for $|v\rangle$, then normalise.

### Worked example 1.6

Take $A = \begin{pmatrix}2&1\\1&2\end{pmatrix}$. Here $\operatorname{tr}A = 4$ and $\det A = 3$, so $\lambda^2 - 4\lambda + 3 = (\lambda-3)(\lambda-1) = 0$. The eigenvalues are $3$ and $1$.

- For $\lambda = 3$: $(A-3I)|v\rangle = \begin{pmatrix}-1&1\\1&-1\end{pmatrix}\begin{pmatrix}x\\y\end{pmatrix} = 0$ gives $x = y$. Normalised: $\frac{1}{\sqrt2}\begin{pmatrix}1\\1\end{pmatrix} = |+\rangle$.
- For $\lambda = 1$: $x = -y$, so the eigenvector is $\frac{1}{\sqrt2}\begin{pmatrix}1\\-1\end{pmatrix} = |-\rangle$.

The two eigenvectors are orthogonal. That always happens for a Hermitian matrix with two different eigenvalues.

### Worked example 1.7

The Pauli matrices each have eigenvalues $+1$ and $-1$. For $Z$ the eigenvectors are $|0\rangle$ (eigenvalue $+1$) and $|1\rangle$ (eigenvalue $-1$). For $X$ they are $|+\rangle$ and $|-\rangle$. These pairs will become the **Z basis** and the **X basis** for measurement in Workbook 2.

> **Note.** An eigenvector is only fixed up to a nonzero factor. If $|v\rangle$ is an eigenvector, so are $2|v\rangle$ and $-|v\rangle$ and $i|v\rangle$. After normalising, the only freedom left is a phase $e^{i\gamma}$. NumPy may return $-|+\rangle$ instead of $|+\rangle$. Both are correct.

```python
A = np.array([[2, 1], [1, 2]])
vals, vecs = np.linalg.eigh(A)      # eigh: for Hermitian matrices, eigenvalues sorted
print("eigenvalues:", vals)
print("eigenvectors (columns):\n", np.round(vecs, 4))
for lam, v in zip(vals, vecs.T):
    print(f"A v = {lam:.0f} v ?", np.allclose(A @ v, lam * v))
```

## 1.8 A first look at tensor products

To describe two qubits together, we combine their vectors with the **tensor product** (also called the Kronecker product), written $\otimes$. For two vectors in $\mathbb{C}^2$:

$$
\begin{pmatrix}a_0\\a_1\end{pmatrix}\otimes\begin{pmatrix}b_0\\b_1\end{pmatrix} = \begin{pmatrix}a_0b_0\\a_0b_1\\a_1b_0\\a_1b_1\end{pmatrix}.
$$

Each entry of the first vector multiplies the whole second vector. The result has $2\times2 = 4$ entries. We write $|0\rangle\otimes|1\rangle$ as $|0\rangle|1\rangle$ or simply $|01\rangle$:

$$
|01\rangle = \begin{pmatrix}1\\0\end{pmatrix}\otimes\begin{pmatrix}0\\1\end{pmatrix} = \begin{pmatrix}0\\1\\0\\0\end{pmatrix}.
$$

For $n$ qubits the vector has $2^n$ entries. Thirty qubits already need about a billion complex numbers to write down in full. This is one reason quantum systems are hard to simulate on ordinary computers. It does *not* mean a quantum computer can read out all those numbers. You will see why in Workbook 2.

## 1.9 Probability you will need

A **probability distribution** over outcomes $x_1,\dots,x_k$ is a list of numbers $p_1,\dots,p_k$ with $p_j \geq 0$ and $\sum_j p_j = 1$.

If each outcome carries a number value $x_j$, the **expected value** (mean) and **variance** are

$$
\mathbb{E}[x] = \sum_j p_j\,x_j, \qquad \operatorname{Var}(x) = \mathbb{E}[x^2] - \big(\mathbb{E}[x]\big)^2 .
$$

### Worked example 1.8

A measurement gives $+1$ with probability $0.3$ and $-1$ with probability $0.7$. Then $\mathbb{E}[x] = 0.3 - 0.7 = -0.4$. Since $x^2 = 1$ always, $\mathbb{E}[x^2] = 1$ and $\operatorname{Var}(x) = 1 - 0.16 = 0.84$.

### Frequencies from repeated runs

Quantum programs are run many times. Each run is called a **shot**. If an outcome has probability $p$ and you run $N$ shots, the number of times you see it is random. The **observed frequency** $\hat p = n/N$ is close to $p$ for large $N$, but not equal to it. Its typical error (the **standard error**) is

$$
\sigma_{\hat p} = \sqrt{\frac{p(1-p)}{N}} .
$$

To halve the error you need four times as many shots. For $p = 0.3$ and $N = 1000$ shots, $\sigma_{\hat p} = \sqrt{0.21/1000} \approx 0.0145$. So frequencies anywhere from about $0.27$ to $0.33$ are normal.

```python
rng = np.random.default_rng(seed=2027)
p = 0.3
for N in [10, 100, 1000, 10000]:
    outcomes = rng.random(N) < p          # True with probability p
    p_hat = outcomes.mean()
    se = np.sqrt(p * (1 - p) / N)
    print(f"N={N:6d}  observed={p_hat:.4f}  true={p}  typical error={se:.4f}")
```
